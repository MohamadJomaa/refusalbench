# Baseline judging with Gemini (no GPU needed)

The baseline replies are already generated. This notebook only **grades** them with Gemini so the baseline can be compared fairly with Extension 2.

**Any Google account works, and a CPU runtime is enough** (Runtime > Change runtime type > CPU). No T4 is needed.

**Once per account:**
1. In Google Drive create `MyDrive/refusalbench-results/` and upload the folder `qwen15_7b_baseline` into it (unzip `baseline/results/qwen15_7b_baseline_judged_day1.zip` on your PC first). It must contain `target_outputs.jsonl`.
2. Colab left bar > key icon > **Add new secret**: name `GEMINI_API_KEY`, paste your key, switch "Notebook access" on.

**Every day:** Runtime > **Run all**. It stops by itself when the free daily limit (500 requests per key) is used up, and prints how many replies are graded. Run it again tomorrow, or put a key from another Google project in the secret and run again now.

**Stay on ONE account for the whole judging**, because the progress (`judged_outputs.jsonl`) lives in that account's Drive. To change account, upload the latest downloaded zip's folder to the new Drive first.

Done when it prints 100% graded. The last step downloads a zip: put it in `baseline/results/` on your PC.

```python
# BASELINE JUDGING - judge-only, NO GPU needed (any Google account, Runtime type: CPU is fine).
# Grades Qwen's baseline replies with Gemini. Safe to re-run every day: finished work is kept and skipped.
import os, subprocess, sys

REPO = 'MohamadJomaa/refusalbench'   # the team repository (private: needs the GITHUB_TOKEN secret)
OUT  = '/content/drive/MyDrive/refusalbench-results'        # the folder qwen15_7b_baseline/ must be inside this folder
RUN  = f'{OUT}/qwen15_7b_baseline'
JUDGE_MODEL = 'gemini-3.5-flash-lite'                       # the pinned judge (must match the config)

# A tiny program that tests the key with ONE call. It is run in a separate process.
KEY_TEST = """
import os
from google import genai
try:
    client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])  # keep a reference, or it closes mid-call
    r = client.models.generate_content(model=os.environ['JUDGE_MODEL'], contents='Reply with the single word OK')
    print('KEY TEST: works.', repr((r.text or '')[:20]))
except Exception as e:
    m = str(e)
    if '429' in m or 'RESOURCE_EXHAUSTED' in m:
        print('KEY TEST: this key has NO quota left today (error 429).')
    elif '400' in m or '403' in m or 'API key' in m:
        print('KEY TEST: the key is not valid: ' + m[:150])
    else:
        print('KEY TEST: unexpected error: ' + m[:200])
"""

KEY_HELP = """
>>> NOT STARTED (nothing was graded). What to do:
    - "NO quota left": this is the OLD key, or a key from the SAME Google project.
      Create the key in a NEW project: aistudio.google.com/apikey > Create API key > "Create API key in new project".
      Put it in the Colab secret named exactly GEMINI_API_KEY (replace the old value), keep "Notebook access" on,
      and run again. The last 4 characters printed above must be the ones of the NEW key.
    - Or wait for the daily reset (tomorrow) and run again with any key.
    - "not valid": the key was pasted incompletely or with a space. Copy it again.
"""


def run(cmd, **kw):
    """Run a command and print its output LIVE."""
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, bufsize=0,
                         env={**os.environ, 'PYTHONUNBUFFERED': '1'}, **kw)
    while True:
        chunk = os.read(p.stdout.fileno(), 4096)
        if not chunk:
            break
        print(chunk.decode('utf-8', 'replace'), end='', flush=True)
    if p.wait() != 0:
        raise RuntimeError(f'command failed (exit {p.returncode}): {cmd}')


def count(path):
    return sum(1 for _ in open(path, encoding='utf-8')) if os.path.exists(path) else 0


def clone_repo(repo, dest, token):
    """Check the token with the GitHub API first, so a bad token gives a clear message, then clone."""
    import subprocess, urllib.request, urllib.error
    req = urllib.request.Request(f'https://api.github.com/repos/{repo}',
                                 headers={'Authorization': f'Bearer {token}', 'Accept': 'application/vnd.github+json'})
    try:
        urllib.request.urlopen(req)
    except urllib.error.HTTPError as e:
        raise SystemExit(f'GitHub refused the GITHUB_TOKEN secret (HTTP {e.code}). ' + {
            401: 'The token is invalid or expired: create a new one and replace the secret.',
            404: 'The token cannot see the repo: give it access to ' + repo + ' with Contents: Read-only.',
            403: 'The token is blocked: check its permissions and its expiry.'}.get(e.code, ''))
    if not os.path.exists(dest):
        url = f'https://x-access-token:{token}@github.com/{repo}.git'
        res = subprocess.run(['git', 'clone', '-q', url, dest], capture_output=True, text=True)
        if res.returncode != 0:
            raise SystemExit('git clone failed: ' + res.stderr.replace(token, '***'))


def main():
    from google.colab import drive
    drive.mount('/content/drive')

    # 1) The saved baseline replies must be on THIS account's Drive
    if not os.path.exists(f'{RUN}/target_outputs.jsonl'):
        print('\n>>> NOT STARTED: the baseline replies are not on this Google account\'s Drive.\n'
              '    Do this once: in Drive create the folder  MyDrive/refusalbench-results  and upload the folder\n'
              '    "qwen15_7b_baseline" into it (unzip baseline/results/qwen15_7b_baseline_judged_day1.zip on your PC first).\n'
              f'    It must contain: {RUN}/target_outputs.jsonl\n    Then run this cell again.')
        return
    print(f'Baseline replies found: {count(f"{RUN}/target_outputs.jsonl")} replies.')

    # 2) The Gemini key (Colab secret named GEMINI_API_KEY)
    try:
        from google.colab import userdata
        os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY')
        print('Gemini key found.')
    except Exception:
        print('\n>>> NOT STARTED: the Colab secret GEMINI_API_KEY was not found.\n'
              '    1. Click the KEY icon in the left bar of Colab.\n'
              '    2. "Add new secret": Name = GEMINI_API_KEY (exactly), Value = your key.\n'
              '    3. Switch ON the "Notebook access" toggle next to it.\n'
              '    4. Run this cell again.\n'
              '    (No key yet? Create a free one at aistudio.google.com > Get API key.)')
        return

    # 3) Code + packages (light: no model is loaded)
    token = ''
    try:
        from google.colab import userdata
        token = userdata.get('GITHUB_TOKEN') or ''          # needed: the team repository is private
    except Exception:
        pass
    clone_repo(REPO, '/content/refusalbench', token.strip())
    os.chdir('/content/refusalbench')
    run(['git', 'pull', '-q'])
    run([sys.executable, '-m', 'pip', 'install', '-q', 'datasets', 'pyyaml', 'google-genai', 'python-dotenv'])

    # 3b) Test the key BEFORE the long run (1 tiny call) and show its last 4 characters, to match it with AI Studio
    key = os.environ['GEMINI_API_KEY']
    print(f'Using a key of {len(key)} characters, ending in ...{key[-4:]}')
    res = subprocess.run([sys.executable, '-c', KEY_TEST], capture_output=True, text=True,
                         env={**os.environ, 'JUDGE_MODEL': JUDGE_MODEL})
    print(res.stdout.strip() or res.stderr.strip()[-300:])
    if 'KEY TEST: works.' not in res.stdout:
        print(KEY_HELP)
        return

    # 4) Judge. --accept-partial = do NOT generate again: score the replies that exist (the 40 that failed earlier are
    #    excluded). It stops by itself when the free daily quota (500 requests per day per key) is used up.
    run([sys.executable, 'baseline/scripts/run_all.py', '--config',
         'baseline/configs/qwen15_7b_baseline.yaml', '--accept-partial', '--out', OUT])

    # 5) How far are we? And a zip to keep on your PC (the Drive copy is already saved)
    n_t, n_j = count(f'{RUN}/target_outputs.jsonl'), count(f'{RUN}/judged_outputs.jsonl')
    print(f'\n=== PROGRESS: {n_j} of {n_t} baseline replies are graded ({n_j / max(n_t, 1):.0%}).')
    print('Done when this reaches 100%. If it stopped on the daily limit: run this cell again tomorrow, or replace the '
          'GEMINI_API_KEY secret with a key from ANOTHER Google project and run again now.')
    try:
        from google.colab import files
        run(f'cd {OUT} && zip -qr /content/qwen15_7b_baseline_judged.zip qwen15_7b_baseline', shell=True)
        files.download('/content/qwen15_7b_baseline_judged.zip')
    except Exception as e:
        print('download skipped:', e)


main()
```
